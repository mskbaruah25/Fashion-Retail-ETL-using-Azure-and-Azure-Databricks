```sql
WITH users AS (
    SELECT *
    FROM ecommerce_fashion.gold.comprehensive_table
    WHERE LOWER(type) = 'user'
)
SELECT
    COUNT(DISTINCT identifierHash) AS total_users,

    COUNT(DISTINCT CASE
        WHEN CAST(productsListed AS DOUBLE) > 0
          OR CAST(productsSold AS DOUBLE) > 0
        THEN identifierHash
    END) AS seller_users,

    COUNT(DISTINCT CASE
        WHEN CAST(productsBought AS DOUBLE) > 0
        THEN identifierHash
    END) AS buyer_users,

    SUM(CAST(productsListed AS BIGINT)) AS total_products_listed,
    SUM(CAST(productsSold AS BIGINT)) AS total_products_sold,
    SUM(CAST(productsBought AS BIGINT)) AS total_products_bought,

    ROUND(
        100.0 * AVG(
            CASE WHEN CAST(hasAnyApp AS BOOLEAN) THEN 1.0 ELSE 0.0 END
        ),
        2
    ) AS app_adoption_pct,

    ROUND(AVG(CAST(seniorityAsYears AS DOUBLE)), 2)
        AS average_account_age_years

FROM users;
```
