```sql
SELECT
    country,
    CAST(sellers AS BIGINT) AS total_sellers,
    CAST(topsellers AS BIGINT) AS top_sellers,

    ROUND(
        100.0 * CAST(
            COALESCE(top_seller_ratio, topsellerratio) AS DOUBLE
        ),
        2
    ) AS top_seller_pct,

    ROUND(
        100.0 * CAST(femalesellersratio AS DOUBLE),
        2
    ) AS female_seller_pct,

    CAST(totalproductslisted AS BIGINT)
        AS total_products_listed,

    CAST(totalproductssold AS BIGINT)
        AS total_products_sold,

    ROUND(
        100.0 * TRY_DIVIDE(
            CAST(totalproductssold AS DOUBLE),
            CAST(totalproductslisted AS DOUBLE)
        ),
        2
    ) AS sell_through_pct,

    ROUND(CAST(meanproductssold AS DOUBLE), 2)
        AS average_products_sold_per_seller,

    ROUND(CAST(meanfollowers AS DOUBLE), 2)
        AS average_seller_followers,

    performance_indicator,
    high_performance,
    activity_level

FROM ecommerce_fashion.gold.comprehensive_table
WHERE sellers IS NOT NULL
ORDER BY total_products_sold DESC;
```
