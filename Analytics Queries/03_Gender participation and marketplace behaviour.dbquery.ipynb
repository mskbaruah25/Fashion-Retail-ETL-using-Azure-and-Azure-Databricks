```sql
WITH gender_summary AS (
    SELECT
        countryCode,
        gender,
        COUNT(DISTINCT identifierHash) AS users,
        SUM(CAST(productsListed AS BIGINT)) AS products_listed,
        SUM(CAST(productsSold AS BIGINT)) AS products_sold,
        SUM(CAST(productsBought AS BIGINT)) AS products_bought,
        AVG(CAST(socialNbFollowers AS DOUBLE)) AS average_followers
    FROM ecommerce_fashion.gold.comprehensive_table
    WHERE LOWER(type) = 'user'
    GROUP BY countryCode, gender
)
SELECT
    countryCode,
    gender,
    users,

    ROUND(
        100.0 * users / NULLIF(SUM(users) OVER (PARTITION BY countryCode), 0),
        2
    ) AS country_user_share_pct,

    products_listed,
    products_sold,
    products_bought,
    ROUND(average_followers, 2) AS average_followers

FROM gender_summary
ORDER BY countryCode, users DESC;
```
