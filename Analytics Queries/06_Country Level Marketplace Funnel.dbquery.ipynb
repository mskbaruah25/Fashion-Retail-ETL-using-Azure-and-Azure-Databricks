```sql
SELECT
    countryCode,
    MAX(country) AS country_name,

    SUM(CAST(socialProductsLiked AS BIGINT)) AS products_liked,
    SUM(CAST(productsWished AS BIGINT)) AS products_wished,
    SUM(CAST(productsBought AS BIGINT)) AS products_bought,

    SUM(CAST(productsListed AS BIGINT)) AS products_listed,
    SUM(CAST(productsSold AS BIGINT)) AS products_sold,

    ROUND(
        100.0 * TRY_DIVIDE(
            SUM(CAST(productsWished AS DOUBLE)),
            SUM(CAST(socialProductsLiked AS DOUBLE))
        ),
        2
    ) AS like_to_wishlist_pct,

    ROUND(
        100.0 * TRY_DIVIDE(
            SUM(CAST(productsBought AS DOUBLE)),
            SUM(CAST(productsWished AS DOUBLE))
        ),
        2
    ) AS wishlist_to_purchase_pct,

    ROUND(
        100.0 * TRY_DIVIDE(
            SUM(CAST(productsSold AS DOUBLE)),
            SUM(CAST(productsListed AS DOUBLE))
        ),
        2
    ) AS listing_sell_through_pct

FROM ecommerce_fashion.gold.comprehensive_table
WHERE LOWER(type) = 'user'
GROUP BY countryCode
ORDER BY products_sold DESC;
```
