```sql
SELECT
    countryCode,
    MAX(country) AS country_name,
    COUNT(DISTINCT identifierHash) AS total_users,

    ROUND(
        100.0 * AVG(
            CASE WHEN CAST(hasAnyApp AS BOOLEAN) THEN 1.0 ELSE 0.0 END
        ),
        2
    ) AS app_adoption_pct,

    ROUND(
        100.0 * AVG(
            CASE WHEN CAST(hasIosApp AS BOOLEAN) THEN 1.0 ELSE 0.0 END
        ),
        2
    ) AS ios_adoption_pct,

    ROUND(
        100.0 * AVG(
            CASE WHEN CAST(hasAndroidApp AS BOOLEAN) THEN 1.0 ELSE 0.0 END
        ),
        2
    ) AS android_adoption_pct,

    ROUND(AVG(CAST(socialNbFollowers AS DOUBLE)), 2)
        AS average_followers,

    ROUND(AVG(CAST(dayssincelastlogin AS DOUBLE)), 2)
        AS average_days_since_login

FROM ecommerce_fashion.gold.comprehensive_table
WHERE LOWER(type) = 'user'
GROUP BY countryCode
ORDER BY total_users DESC;
```
