```sql
WITH users AS (
    SELECT
        *,

        CASE
            WHEN CAST(hasIosApp AS BOOLEAN)
             AND CAST(hasAndroidApp AS BOOLEAN)
                THEN 'iOS and Android'
            WHEN CAST(hasIosApp AS BOOLEAN)
                THEN 'iOS'
            WHEN CAST(hasAndroidApp AS BOOLEAN)
                THEN 'Android'
            WHEN CAST(hasAnyApp AS BOOLEAN)
                THEN 'Other App'
            ELSE 'No App'
        END AS app_platform

    FROM ecommerce_fashion.gold.comprehensive_table
    WHERE LOWER(type) = 'user'
)
SELECT
    app_platform,
    COUNT(DISTINCT identifierHash) AS users,

    ROUND(AVG(CAST(dayssincelastlogin AS DOUBLE)), 2)
        AS average_days_since_login,

    ROUND(AVG(CAST(socialNbFollowers AS DOUBLE)), 2)
        AS average_followers,

    ROUND(AVG(CAST(productsListed AS DOUBLE)), 2)
        AS average_products_listed,

    ROUND(AVG(CAST(productsSold AS DOUBLE)), 2)
        AS average_products_sold,

    ROUND(AVG(CAST(productsBought AS DOUBLE)), 2)
        AS average_products_bought,

    ROUND(AVG(CAST(productsWished AS DOUBLE)), 2)
        AS average_products_wished

FROM users
GROUP BY app_platform
ORDER BY users DESC;
```
